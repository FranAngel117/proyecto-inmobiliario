```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

%matplotlib inline
```

```python
df_casas = pd.read_csv(
    "../data/2023-03-08 Precios Casas RM.csv",
    encoding="latin1",
    sep=";"
)

df_propiedades = pd.read_csv(
    "../data/2023-07-18 Propiedades Web Scrape.csv",
    encoding="latin1",
    sep=None,
    engine="python"
)

print("Precios Casas RM:", df_casas.shape)
print("Propiedades Web Scrape:", df_propiedades.shape)
```

```python
print("=== PRECIOS CASAS RM ===")
display(df_casas.head())

print("\n=== PROPIEDADES WEB SCRAPE ===")
display(df_propiedades.head())

print("\n=== COLUMNAS ===")
print(df_casas.columns.tolist())

print("\n=== TIPOS DE DATOS ===")
display(df_casas.dtypes)
```

```python
print("=== PRECIOS CASAS RM ===")
display(df_casas.describe())

print("\n=== PROPIEDADES WEB SCRAPE ===")
display(df_propiedades.describe())
```

```python
print("=== VALORES FALTANTES: PRECIOS CASAS RM ===")
display(df_casas.isna().sum())

print("\n=== VALORES FALTANTES: PROPIEDADES WEB SCRAPE ===")
display(df_propiedades.isna().sum())
```

```python
print("=== COMUNAS EN PRECIOS CASAS RM ===")
print("Cantidad de comunas:", df_casas["Comuna"].nunique())
print(df_casas["Comuna"].value_counts())

print("\n=== COMUNAS EN PROPIEDADES WEB SCRAPE ===")
print("Cantidad de comunas:", df_propiedades["Comuna"].nunique())
print(df_propiedades["Comuna"].value_counts())
```
